```python
%sql
CREATE OR REPLACE TABLE learning.silver.schema_demo (
    customer_id INT,
    name STRING
) USING DELTA;

INSERT INTO learning.silver.schema_demo
VALUES
  (1, 'John'),
  (2, 'Jane'),
  (3, 'Bob');

  SELECT * FROM learning.silver.schema_demo
  ORDER BY customer_id;
```

```python
from pyspark.sql import functions as F

new_customers = spark.createDataFrame([
(3, "Sara", "Austria"),
(4, "John", "USA")
],
"customer_id INT, name STRING, country STRING"
)

new_customers.printSchema()
display(new_customers)
```

```python
(new_customers.write
  .format("delta")
  .mode("append")
  .saveAsTable("learning.silver.schema_demo"))
```

```python
(new_customers.write
  .format("delta")
  .mode("append")
  .option("mergeSchema", "true")
  .saveAsTable("learning.silver.schema_demo"))
```
